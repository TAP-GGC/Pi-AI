3. Build your own agent

```python

  # --- NAME ---
  # What is your agent called?
agent_name = "INSERT NAME HERE"

  # --- BEHAVIOR ---
  # How should your agent act and communicate?
  # Example: "Be friendly and encouraging" / "Be funny and sarcastic" / "Be calm and professional"
agent_behavior = "INSERT BEHAVIOR HERE"

  # --- RULES ---
  # What must your agent always or never do?
  # Example: "Always end with a fun fact" / "Never give direct answers, only hints"
rule_1 = "INSERT RULE HERE"
rule_2 = "INSERT RULE HERE"
rule_3 = "INSERT RULE HERE"



SYSTEM_PROMPT = f"""
  You are {agent_name}.

  Behavior:
  {agent_behavior}

  Rules you must follow:
  1. {rule_1}
  2. {rule_2}
  3. {rule_3}
  """

print(f"Agent '{agent_name}' is ready!")

```

4. Chat with Agent

```python
from IPython.display import display, HTML, clear_output
import re

chat_history = []

def format_text(text):
    # Convert markdown to HTML
    text = re.sub(r'\*\*\*(.*?)\*\*\*', r'<b><i>\1</i></b>', text)
    text = re.sub(r'\*\*(.*?)\*\*', r'<b>\1</b>', text)
    text = re.sub(r'\*(.*?)\*', r'<i>\1</i>', text)
    text = text.replace('\n', '<br>')
    return text

def chat(user_message):
    contents = [
        {"role": "user", "parts": [{"text": SYSTEM_PROMPT}]},
        {"role": "model", "parts": [{"text": f"Got it! I am {agent_name}. How can I help?"}]},
    ]

    for entry in chat_history:
        contents.append(entry)

    contents.append({"role": "user", "parts": [{"text": user_message}]})

    response = client.models.generate_content(model=MODEL, contents=contents)
    reply = response.text.strip()

    chat_history.append({"role": "user", "parts": [{"text": user_message}]})
    chat_history.append({"role": "model", "parts": [{"text": reply}]})

    # Clear and redisplay the full chat history
    clear_output(wait=True)

    history_html = ""
    for entry in chat_history:
        role = entry["role"]
        text = format_text(entry["parts"][0]["text"])
        if role == "user":
            history_html += f"""
            <div style="margin:8px 0; text-align:right;">
                <span style="background:#4A90D9; color:white; padding:8px 14px; border-radius:16px; display:inline-block; max-width:75%;">
                    {text}
                </span>
                <div style="font-size:11px; color:gray;">You</div>
            </div>"""
        else:
            history_html += f"""
            <div style="margin:8px 0; text-align:left;">
                <span style="background:#F1F1F1; color:black; padding:8px 14px; border-radius:16px; display:inline-block; max-width:75%;">
                    {text}
                </span>
                <div style="font-size:11px; color:gray;">{agent_name}</div>
            </div>"""

    display(HTML(f'<div style="font-family:sans-serif; padding:10px;">{history_html}</div>'))

# Test your agent!
chat("INSERT YOUR FIRST MESSAGE HERE")
```
